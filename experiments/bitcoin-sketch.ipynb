```python

ONE_HASH_DIFF = 0xFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFF 
DIFF_ONE  = 0x00000000FFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFF

def diff_to_target(diff: int) -> int:
    return DIFF_ONE // diff

def target_to_str(target: int) -> str:
    return f"0x{target:064X}"

def diff_to_hashes(diff: int) -> int:
    return ONE_HASH_DIFF // diff_to_target(diff)

def diff_to_hps(diff: int) -> int:
    # divide by block time
    return diff_to_hashes(diff) // 600
```
